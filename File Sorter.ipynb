## Automatic File Sorter in File Explorer

```python
import os, shutil # The shutil module in Python provides a collection of high-level operations for copying, moving, removing, and archiving files and directories.
```

```python
path = r"C:path/to/Sort/" # change from \ to / . need to have / at the end so that it creates the folders in the sort folder
```

```python
file_name = os.listdir(path)
file_name
```

```python
folder_names = ['media files', 'text files', 'pdf files']

for loop in range(0,3):
    if not os.path.exists(path + folder_names[loop]) :
        print(path + folder_names[loop])
        os.makedirs(path + folder_names[loop])
```

```python
for file in file_name:
    if file.lower().endswith((".docx", ".csv")) and not os.path.exists(path + "text files/" + file):
        shutil.move(path + file, path + "text files/" + file)
    elif file.lower().endswith((".jpg", ".png", ".mp4")) and not os.path.exists(path + "media files/" + file):
        shutil.move(path + file, path + "media files/" + file)
    elif file.lower().endswith(".pdf") and not os.path.exists(path + "pdf files/" + file):
        shutil.move(path + file, path + "pdf files/" + file)
    else:
         print(f"{file} was not moved!")
```
